# Cave Set to LaTeX Table Generator

This notebook generates LaTeX tables from cave set summary data.

```python
import json
import os
from pathlib import Path
import pandas as pd
```

```python
# Specify the cave set path
caveset_path = "/home/ga53voq/master_thesis/new_caves/goal"  # Change this to your desired cave set path
print(f"Using cave set: {caveset_path}")
```

```python
# Generate LaTeX table from cave set summary
def generate_caveset_latex_table(caveset_path):
    """
    Generate a LaTeX table from a cave set summary.
    
    Args:
        caveset_path (str): Path to the cave set directory containing caves_summary.json
    
    Returns:
        str: LaTeX table string
    """
    
    # Load caves summary
    caves_summary_path = os.path.join(caveset_path, "caves_summary.json")
    
    if not os.path.exists(caves_summary_path):
        print(f"Error: Could not find caves_summary.json at {caves_summary_path}")
        return None
    
    print(f"Loading cave summary from: {caves_summary_path}")
    
    with open(caves_summary_path, 'r') as f:
        caves_summary = json.load(f)
    
    # Extract cave data
    caves_data = []
    for cave_info in caves_summary.get('caves', []):
        caves_data.append({
            'Cave ID': cave_info.get('id', 'N/A'),
            'Curve Intensity': cave_info.get('curve_intensity', 0.0),
            'Num Dents': cave_info.get('num_dents', 0),
            'Offset Magnitude': cave_info.get('offset_magnitude', 0.0)
        })
    
    # Filter for cave IDs 271 to 300
    caves_data = [cave for cave in caves_data if 271 <= cave['Cave ID'] <= 300]
    
    # Sort by Cave ID
    caves_data.sort(key=lambda x: x['Cave ID'])
    
    print(f"Found {len(caves_data)} caves in the summary (filtered for IDs 271-300)")
    
    # Create DataFrame for easier handling
    df = pd.DataFrame(caves_data)
    
    # Generate LaTeX table
    latex_table = "\\begin{table}[h!]\n"
    latex_table += "\\centering\n"
    latex_table += "\\label{tab:cave_characteristics}\n"
    latex_table += "\\begin{tabular}{cccc}\n"
    latex_table += "\\toprule\n"
    latex_table += "Cave ID & Curve Intensity & Num Dents & Offset Magnitude \\\\\n"
    latex_table += "\\midrule\n"
    
    # Add data rows
    for _, row in df.iterrows():
        cave_id = row['Cave ID']
        curve_intensity = f"{row['Curve Intensity']:.3f}"
        num_dents = int(row['Num Dents'])
        offset_magnitude = f"{row['Offset Magnitude']:.3f}"
        
        latex_table += f"{cave_id} & {curve_intensity} & {num_dents} & {offset_magnitude} \\\\\n"
        latex_table += "\\bottomrule\n"
    
    latex_table += "\\end{tabular}\n"
    latex_table += "\\caption{Cave Set Characteristics}\n"
    latex_table += "\\end{table}\n"
    
    return latex_table, df

# Generate the table
latex_table, cave_df = generate_caveset_latex_table(caveset_path)

if latex_table:
    print("\\n" + "="*60)
    print("LATEX TABLE OUTPUT")
    print("="*60)
    print(latex_table)
    
    # Display summary statistics
    print("\\n" + "="*60)
    print("CAVE SET SUMMARY STATISTICS")
    print("="*60)
    print(f"Total caves: {len(cave_df)}")
    print(f"Curve Intensity - Range: [{cave_df['Curve Intensity'].min():.3f}, {cave_df['Curve Intensity'].max():.3f}], Mean: {cave_df['Curve Intensity'].mean():.3f}")
    print(f"Num Dents - Range: [{cave_df['Num Dents'].min()}, {cave_df['Num Dents'].max()}], Mean: {cave_df['Num Dents'].mean():.1f}")
    print(f"Offset Magnitude - Range: [{cave_df['Offset Magnitude'].min():.3f}, {cave_df['Offset Magnitude'].max():.3f}], Mean: {cave_df['Offset Magnitude'].mean():.3f}")
    
    # Display first few rows for verification
    print("\\n" + "="*60)
    print("FIRST 10 CAVES (for verification)")
    print("="*60)
    print(cave_df.head(10).to_string(index=False, float_format='%.3f'))
    
    # Save to file
    output_file = Path(caveset_path) / "cave_characteristics_table.tex"
    with open(output_file, 'w') as f:
        f.write(latex_table)
    
    print(f"\\nLaTeX table saved to: {output_file}")
    print("="*60)
else:
    print("Failed to generate LaTeX table. Please check the cave set path and summary file.")
```
