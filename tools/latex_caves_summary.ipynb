```python
import json
import os

def generate_latex_table():
    new_caves_folder = "/home/ga53voq/master_thesis/new_caves"
    table_rows = []
    
    # Get all cave folders
    cave_folders = [f for f in os.listdir(new_caves_folder) if os.path.isdir(os.path.join(new_caves_folder, f))]
    cave_folders.sort(key=lambda x: int(x.replace('cave_', '')))  # Sort by cave number
    
    for cave_folder in cave_folders:
        config_path = os.path.join(new_caves_folder, cave_folder, "cave_config.json")
        
        if os.path.exists(config_path):
            with open(config_path, 'r') as f:
                config = json.load(f)
            
            # Extract values from generation section
            cave_id = config["generation"]["cave_id"]
            curve_intensity = config["generation"]["curve_intensity"]
            seed = config["generation"]["seed"]
            
            # Extract values from mesh sections
            vertex_count = config["original_mesh"]["vertex_count"]
            face_count = config["original_mesh"]["face_count"]
            voxel_count = config["voxelization"]["voxel_count"]
            
            # Create table row
            row = f"{cave_id} & {curve_intensity} & {seed} & {vertex_count} & {face_count} & {voxel_count} \\\\"
            table_rows.append(row)
    
    # Create complete table
    header = """\\begin{table}[htbp]
\\centering
\\begin{tabular}{|c|c|c|c|c|c|}
\\hline
Cave ID & Curve Intensity & Seed & Vertices & Faces & Voxels \\\\
\\hline"""
    
    footer = """\\hline
\\end{tabular}
\\caption{Cave Configuration Parameters and Mesh Statistics}
\\label{tab:cave_configs}
\\end{table}"""
    
    latex_table = header + "\n" + "\n".join(table_rows) + "\n" + footer
    
    print("LaTeX table ready to copy:")
    print("=" * 80)
    print(latex_table)
    print("=" * 80)
    
    return latex_table

# Generate the table
latex_output = generate_latex_table()
```
